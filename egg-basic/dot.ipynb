# DOT Styles
Different Showing Styles

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

```julia
float_ = linearwfsa("f l o a t")
if_ = linearwfsa("i f")
int_ = linearwfsa("i n t")
else_ = linearwfsa("e l s e")
end_ = linearwfsa("e n d")
T = float_ + if_ + int_ + else_ + end_
R = noeps(T)
D = reorder(det(R))
M = min(D)
arcline = "spline";
setdpi(60)
```

normal show of a dot

```julia
showfst(M, keepsame=false, splines=arcline)
```

all nodes in one line

```julia
pshow(linedot(M, keepsame=false, splines=arcline))
```

show nodes according to its topology order

```julia
pshow(leveldot(M, keepsame=false, splines=arcline))
```

show nodes according to its symbol depth from start state

```julia
pshow(depthdot(M, "fwd", keepsame=false, splines=arcline))
```

show nodes according to its symbol depth from final state

```julia
pshow(depthdot(M, "bwd", keepsame=false, splines=arcline))
```

```julia
showfsts([R, D, M], labels=["T", "Determined T", "Minimized T"], keepsame=false, splines=arcline)
```
