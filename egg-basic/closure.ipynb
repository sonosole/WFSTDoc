# Closure and Star
$$T^*(x,y) = \underset{n=0}{\overset{∞}{\Huge ⊕}} T^n(x,y)  \texttt { \gray{ closure } }$$
$$T^+(x,y) = \underset{n=1}{\overset{∞}{\Huge ⊕}} T^n(x,y)  \texttt { \gray{ star } } $$

Use semiring pkg and showing pkg

```julia
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

a example from *OpenFst: a General and Efficient Weighted Finite-State Transducer Library Part I. Library Design and Use*, page 19

```julia
let W = Dioid{max,+,Float64}
    T = WFST{Int, String, String, W}();
    addarc(T, "0 1 green:green/0.4")
    addarc(T, "0 2 blue:blue/1.2")
    addstart(T, 0)
    addfinal(T, 1, 0.0)
    addfinal(T, 2, 0.3)
    setdpi(50)
    showfsts([T, star(T), plus(T)], 
            labels = ["T", "star(T)","plus(T)"],
            keepsame = false,
            rankdir = "LR",
            boxcolor = "grey")
end
```
