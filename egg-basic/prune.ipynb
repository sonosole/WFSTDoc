## Prune by Weight 🏋️

### API
```julia
wprune(T::WFST{N,I,O,W}, θ::Real; iskept::Function = ≤;
                                    atol::Real = 0,
                                    rtol::Real = atol>0 ? 0 : 1e-3) where {N,I,O,W}

  Deletes states and arcs in T that do not belong to a successful 😍 path. An arc is
  removed if its score S is worse than t = Sᵇ ⊗  W(θ) where Sᵇ is the score of the best
  path. So the arc whose score is better than Sᵇ ⊗  W(θ) is kept. In the semantic of
  cost, iskept usually means ≤ t, as for probabilities or likelyhood, iskept usually
  means > t, e.g.

    •  when score is cost 💳 , if S ≤ Sᵇ ⊗  W(θ), then kept.
    •  when score is probability 🎲 , if S > Sᵇ ⊗  W(θ), then kept.

  atol and rtol are used to control the tolerance of error when calculating the
  potentials.
```

now try pruning in log semiring

````julia
include("d://repos//FSM.jl//src//FSM.jl")
let WT = Dioid{min,+,Float64}
    am = fweight(x -> -log(x), amfst(4,3))
    am = convert(WFST{Int,String,String,WT}, am)
    AM1 = deepcopy(am);
    AM2 = deepcopy(am);

    trws = paths(AM1, 9)
    TOP1 = last(trws[1])
    TOP9 = last(trws[9])

    t = trunc(value(TOP9 / TOP1), digits=1)
    P1 = wprune(AM1, t, iskept= <);
    P2 = wprune(AM2, t, iskept= ≥);
    setdpi(60)
    showfsts([am, P1, P2],
             digits=3,
             boxcolor = "grey",
             labels=["origin acoustic scores",
                     "Threshold <$t, low costs kept",
                     "Threshold ≥$t, high costs kept"])
end
````

in a probabilistic semiring

````julia
let timesteps=5, kstates=3
    am  = amfst(timesteps, kstates)
    AM1 = deepcopy(am);
    AM2 = deepcopy(am);
    t = 0.2
    P1 = wprune(AM1, t, iskept= ≥);
    P2 = wprune(AM2, t, iskept= <);
    setdpi(60)
    showfsts([am, P1, P2], 
             labels=["origin acoustic probabilities", 
                     "Threshold ≥$t, high probabilities kept", 
                     "Threshold <$t, low probabilities kept"])
end
````

## prune by Top-k 🥇🥈🥉🏆
### API
```julia
 kprune(T::WFST{N,I,O,W}, k::Int; iskept::Function=(<),
                                    atol::Real=0,
                                    rtol::Real=atol>0 ? 0 : 1e-3) where {N,I,O,W}

  Deletes states and arcs in `T` that do not belong to top-k paths.

    •  when score is cost / loss 💳 , then smaller is better, iskept is < or ≤.
    •  when score is probability 🎲 , then bigger is better, iskept is > or ≥.

  atol and rtol are used to control the tolerance of error when calculating the potentials.
  ```

````julia
let WT = Dioid{min,+,Float64}
    am = fweight(x -> -log(x), demofst(3,5))
    am = convert(WFST{Int,String,String,WT}, am)
    
    # top 3 paths
    println("├─────────── top 3")
    wtr = paths(am, 3, cmpf = <);
    for (i,(tr, w)) ∈ enumerate(wtr)
        print(i, " weight=$w, path:")
        printpath(tr)
    end

    # the same top 3 paths
    println("├─────────── top 3")
    WTR = topk(am, 3, iskept = <);
    for (i,(tr, w)) ∈ enumerate(WTR)
        println(i, " weight=$w, path:", trail(tr))
    end

    TOP2 = kprune(deepcopy(am), 2, iskept = <)
    TOP4 = kprune(deepcopy(am), 4, iskept = <)
    setdpi(50)
    showfsts([am, TOP2, TOP4], labels=["Origin", "top-2", "top-4"], digits=1, boxcolor="gray")
end
````
