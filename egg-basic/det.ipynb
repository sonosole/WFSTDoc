# Determinization
**Principle**: Weighted determinization — converts a non-deterministic WFST into an equivalent deterministic WFST, where each state has at most one leaving arc per input symbol. **Not** all WFSTs are determinizable (e.g. those with non-converging epsilon cycles).

**Purpose**: Eliminates ambiguity, reduces state count, and is a prerequisite for minimization, complementation and other operations.

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using Symbolics
using WeightedFST

showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

a simple example

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int,String,String,W}()
    addarc(T, "0 1 a:X/0.5")
    addarc(T, "0 2 a:Y/1.2")
    addarc(T, "1 4 b:")
    addarc(T, "1 5 c:Z")
    addarc(T, "2 6 b:/1")
    addarc(T, "2 6 d:Z")
    addarc(T, "2 6 d:")
    addarc(T, "2 7 b:")
    addarc(T, "7 8 f:")

    addstart(T, 0, 0.2)
    addfinal(T, 4, 0.3)
    addfinal(T, 5, 0.1)
    addfinal(T, 6, 0.0)
    addfinal(T, 8, 0.0)
    setdpi(50)
    D = det(deepcopy(T))
    showfsts([T, D], 
             labels=["T", "det(T)"],
             keepone=true)
end
```

another simple example

```julia
let
    T = WFST{Int,String,String,Float64}()
    addarc(T, "0 1 d:x/0.4")
    addarc(T, "0 1 d:y/0.1")
    addarc(T, "0 1 d:y/0.5")
    addarc(T, "1 2 e:z")
    addstart(T, 0)
    addfinal(T, 2)
    showfsts([T, tdet(deepcopy(T)), det(deepcopy(T))], 
             labels=["T", "tdet(T)", "det(T)"],
             boxcolor="lightgrey")
end
```

homophonic words

```julia
let
    T = dictwfst("night n ay t") + 
        dictwfst("knight n ay t")
    T = noeps(T)
    D1 = T |> tdet
    D2 = T |> det
    
    showfsts([T, D1, D2, reorder(D2)], labels=["T", "tdet(T)", "det(T)", "reorder(det(T))"],)
end
```

weight-pushing towards initial states, sum of weights of leaving arcs is one.

```julia
let
    char_ = dictwfst("char c h a r")
    const_ = dictwfst("const c o n s t")
    continue_ = dictwfst("continue c o n t i n u e")
    if_ = dictwfst("if i f")
    int_ = dictwfst("int i n t")
    else_ = dictwfst("else e l s e")
    end_ = dictwfst("end e n d")
    short_ = dictwfst("short s h o r t")
    sign_ = dictwfst("signed s i g n e d")
    sizeof_ = dictwfst("sizeof s i z e o f")
    T = noeps(char_ + const_ + continue_ + if_ + int_ + else_ + end_ + short_ + sign_ + sizeof_)
    M = T |> detwfst |> reorder
    showfst(M, keepsame=false, digits=1)
end
```

weight-pushing towards final states, sum of weights of incoming arcs is one.

```julia
let
    char_ = dictwfst("char c h a r")
    const_ = dictwfst("const c o n s t")
    continue_ = dictwfst("continue c o n t i n u e")
    if_ = dictwfst("if i f")
    int_ = dictwfst("int i n t")
    else_ = dictwfst("else e l s e")
    end_ = dictwfst("end e n d")
    short_ = dictwfst("short s h o r t")
    sign_ = dictwfst("signed s i g n e d")
    sizeof_ = dictwfst("sizeof s i z e o f")
    T = noeps(char_ + const_ + continue_ + if_ + int_ + else_ + end_ + short_ + sign_ + sizeof_)
    M = T |> reverse |> detwfst |> reverse |> reorder
    showfst(M, keepsame=false,digits=3)
end
```

inspect determination princple by symbols

```julia
begin
    @variables a v y₁ y₂ y₃ w1 w2 w3
    T = WFST{String, String, Num, Num}();
    addstart(T, "start")

    addarc(T, "start", "p", "i", a, v)
    addarc(T, "p", "q1", "x", y₁, w1)
    addarc(T, "p", "q1", "x", y₂, w2)
    addarc(T, "p", "q2", "x", y₃, w3)
    R = detwfst(T)
    showfsts([T,R], labels=["T", "det(T)"], keepone=true)
end
```
