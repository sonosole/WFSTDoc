# Difference $X - Y$ and Complement $^¬Y$

**Principle**: Outputs the set of strings accepted by X but not by Y. Implemented via complement and intersection: $X - Y = X \cap \text{complement}(Y)$. 

**Purpose**: Filtering illegal sequences, excluding existing patterns, etc.

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

example from *OpenFst:  a General and Efficient Weighted Finite-State Transducer Library Part I. Library Design and Use*, page 29

```julia
let
    G = WFST()
    addpath(G, "0 red/0.5 0 green/0.3 1 blue/0.0 2")
    addarc(G, "1 2 yellow/0.6")
    addstart(G, 0)
    addfinal(G, 2, 0.8)

    T = WFST()
    addpath(T, "0 red 1 green 1 yellow 2")
    addarc(T, "1 0 blue")
    addstart(T, 0)
    addfinal(T, 0)
    addfinal(T, 2)
    T̄ = complement(T)
    setdpi(50)
    showfsts([G, T, connect(G-T), T̄],
             labels = ["G", "T", "G - T", "T̄  or  ¬T"],
             boxcolor="lightgray",
             keepsame=false)
end
```

and an example to remove dirty words

```julia
let SAVEDIR = pwd()
    G = WFST()
    addpath(G, "0 FUCK/0.5 1 You/0.3 2")
    addpath(G, "0 Hello/0.1 1 Asshole/0.3 2")
    addstart(G, 0)
    addfinal(G, 2, 0.7)

    T = WFST()
    addpath(T, "0 FUCK 1 You 2")
    addstart(T, 0)
    addfinal(T, 0)
    addfinal(T, 2)
    T̄ = complement(T, ilabelsof(G))
    BLKS = repeat(" ", 70)
    showfsts([G, T, connect(G-T), T̄],
             labels=["G", "T", "G - T", "T̄=complement(T)"*BLKS],
             boxcolor="lightgray",
             keepsame=false)
end
```
