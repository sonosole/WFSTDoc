# Epsilon-free Composation
$X ∘ Y$ Producing a new WFST whose input is the input of $X$ and output is the output of $Y$, with weights equal to the semiring product of the two weights.

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

a simple example

```julia
let 
    a = WFST();
    addarc(a, 0, 1, "cat",    "cats",    0.1)
    addarc(a, 1, 3, "cute",   "lovely",  0.5)
    addarc(a, 0, 2, "dog",    "dogs",    0.2)
    addarc(a, 2, 3, "eating", "barking", 0.6)
    addstart(a, 0, 0.1)
    addfinal(a, 3, 0.9)

    # Upper case for first letter
    b = WFST();
    addarc(b, 0, 0, "cats",    "Cats",    1.0)
    addarc(b, 0, 0, "lovely",  "Lovely",  1.0)
    addarc(b, 0, 0, "dogs",    "Dogs",    1.0)
    addarc(b, 0, 0, "barking", "Barking", 1.0)
    addstart(b, 0, 1.9)
    addfinal(b, 0, 1.1)

    c = compose(a,b)
    setdpi(50)
    showfsts([a,b,c], 
             labels=["a", "b", "compose(a,b)"],
             rankdir="LR")
end
```

a example from two papers: 
+ *Mehryar Mohri, Weighted Finite-State Transducer Algorithms An Overview*, Figure 1
+ *Mehryar Mohri, Weighted Automata Algorithms* : Fig. 6

```julia
let
    T = WFST();
    addarc(T, "0 1 a:b/0.1")
    addarc(T, "1 0 a:b/0.2")
    addarc(T, "1 2 b:b/0.3")
    addarc(T, "1 3 b:b/0.4")
    addarc(T, "2 3 a:b/0.5")
    addarc(T, "3 3 a:a/0.6")
    addstart(T, 0, 0.0)
    addfinal(T, 3, 0.7)

    G = WFST();
    addarc(G, "0 1 b:b/0.1")
    addarc(G, "1 1 b:a/0.2")
    addarc(G, "1 2 a:b/0.3")
    addarc(G, "1 3 a:b/0.4")
    addarc(G, "2 3 b:a/0.5")
    addstart(G, 0, 0.0)
    addfinal(G, 3, 0.6)
    setdpi(50)
    showfsts([T,G, compose(T,G)],
             labels=["T","G", "compose(T, G)"])
end
```

# Compose with Epsilon
When there is any epsilon in the two WFSTs, the above ε-free compositon fails. Let's see a char-to-word FST example

```julia
begin # a char to lowercase words 
    char_ = dictwfst("char c h a r")
    const_ = dictwfst("const c o n s t")
    continue_ = dictwfst("continue c o n t i n u e")
    if_ = dictwfst("if i f")
    int_ = dictwfst("int i n t")
    else_ = dictwfst("else e l s e")
    end_ = dictwfst("end e n d")
    short_ = dictwfst("short s h o r t")
    sign_ = dictwfst("signed s i g n e d")
    sizeof_ = dictwfst("sizeof s i z e o f")
    T = noeps(char_ + const_ + continue_ + if_ + int_ + else_ + end_ + short_ + sign_ + sizeof_)
    showfst(min(T))
end
```

then we create a WFST that converts lower-case words to upper-case-first words

```julia
begin
    G = WFST()
    addstart(G, 0)
    addfinal(G, 0)
    for str ∈ string.(split("char const continue if int else end short signed sizeof"))
        addarc(G, 0, 0, str, uppercasefirst(str))
    end
    showfst(G, rankdir="LR")
end
```

we first try to compose those two

```julia
showfst(compose(T, G), rankdir="TB") # ε-free composition fails
```

but in vain we can't get a corret result as we expected. Finally, let's try a composition with epsilon matching filter to make an expected result

```julia
Φ = EpsMatcher("ϵ1", "ϵ2", Float64)   # epsilon matching filter
C = Φ(T, G)                           # or call compose(T, G, Φ), this works
showfst( min(C) )
```
