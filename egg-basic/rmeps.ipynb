# $ε$ Removal
**Principle**: Eliminates all epsilon edges (edges where both input and output are $\varepsilon$) from the WFST, yielding an equivalent epsilon-free WFST.

**Purpose**: Simplifies WFST structure, improves efficiency of determinization, composition and other operations.

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

A epsilon removal demo, the following input WFST is from paper *Generic ϵ-Removal and Input ϵ-Normalization Algorithms for Weighted Transducers*

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int,String,String,W}()
    addpath(T, "0 a/0.1 1 :/0.2 2 :/0.8 4")
    addpath(T, "2 a/0.5 4")
    addpath(T, "2 :/0.4 3 :/0.1 2")
    addpath(T, "3 :/0.9 1")
    addpath(T, "2 b/0.6 1")
    addpath(T, "1 b/0.3 4")
    addpath(T, "2 c/0.7 0")
    addstart(T, 0)
    addfinal(T, 4, 0.2)
    CR = connect(rmeps(T))
    RCRR = reverse(connect(rmeps(reverse(T))))
    setdpi(50)
    showfsts([T, CR, RCRR],
             labels=["T", "rmeps(T)", "reverse(rmeps(reverse(T)))"],
             boxcolor="grey",
             keepsame=false)
end
```

a linear WFST

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int,String,String,W}()
    addpath(T, "0 /0.1 1 /0.2 2 /0.3 3 a/1.0 4")
    addstart(T, 0)
    addfinal(T, 4, 1.0)
    
    showfsts([T, noeps(T)],
             labels=["T","noeps(T)"], 
             keepsame=false)
end
```

```julia
let W = Dioid{min,+,Float64}

    T = WFST{Int, String, String, W}()
    addpath(T, "0 a:b/0.1 1 :/0.2 2 :/0.3 3")
    addpath(T, "1 :/0.3 3")
    addstart(T, 0)
    addfinal(T, 3, 0.5)

    showfsts([T, noeps(T)],
             labels = ["T", "ε-removed of T"],
             boxcolor="grey",
             bgcolor="white")
end
```

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int,String,String,W}()
    addstart(T, 0)
    addfinal(T, 2, 0.4)
    addpath(T, "0 : 1 : 2 : 0")
    addpath(T, "0 a 1 b 2")
    x = deepcopy(T)
    showfsts([T, rmeps(x)], 
             labels=["T", "rmeps(T)"], keepone=false, keepsame=false)
end
```

how about using symbols ?

```julia
using Symbolics
```

```julia
let
    @variables a b c d e f g
    T = WFST{Int, String, String, Num}();
    addstart(T, 0)
    addeps(T, 0,1, b)
    addeps(T, 1,2, c)
    addarc(T, 2,3, "X","X", d)
    addarc(T, 2,4, "Y","Y", e)
    addarc(T, 1,2, "Z","Z", g)
    addeps(T, 2,5, f)
    addfinal(T, 3)
    addfinal(T, 4)
    addfinal(T, 5)
    R = noeps(T)
    showfsts([T,R],labels=["Origin T", "ε-removed of T"])
end
```

```julia
let
    @variables a b c d e f g
    T = WFST{Int, String, String, Num}();
    addstart(T, 0, a)
    addeps(T, 0,1, b)
    addeps(T, 1,2, c)
    addarc(T, 2,3, "X","X", d)
    addeps(T, 2,4, e)
    addfinal(T, 3, f)
    addfinal(T, 4, g)
    R = noeps(T)
    showfsts([T,R],labels = ["T","rmeps(T)"])
end
```
