# Input and Output $ε$ Removal

```julia
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
setdpi(50)
```

A epsilon removal demo, the following input WFST is from paper: *Generic ϵ-Removal and Input ϵ-Normalization Algorithms for Weighted Transducers*

```julia
let W = Dioid{min,+,Float64}

    Txy = WFST{Int, String, String, W}()
    addarc(Txy, "0 4 d/0.5")
    addpath(Txy, "0 a:b/0.1 1 :b/0.2 2 :c/0.3 3 b:d/0.4 4")
    addpath(Txy, "1 :f/0.3 3 f:d/0.6 4")
    addstart(Txy, 0)
    addfinal(Txy, 4, 0.4)
    Tyx = swapio(Txy)

    @assert swapio(rmoeps(Tyx)) == rmieps(Txy)
    
    showfsts([Txy, rmieps(Txy), Tyx, rmoeps(Tyx)],
             labels = ["Txy", "input ε removal of Txy", "Tyx", "output ε removal of Tyx"],
             boxcolor = "grey",
             bgcolor = "transparent")
end
```
