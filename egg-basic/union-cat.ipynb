# Union(+) and Concat(*)

$$
(T_1 ⊕ T_2)(x,y) = T_1(x,y) ⊕ T_2(x,y)
$$

$$
(T_1 ⊙ T_2)(x,y) = \underset{x = x_1 * x_2, y= y_1 * y_2}{\Huge ⊕} T_1(x_1,y_2) ⊙ T_2(x_2,y_2)
$$

```julia
using Dioids
using ShowDOT
using WeightedFST
showfsts = pshow ∘ dots;
```

```julia
# Speech Recognition Algorithms Using Weighted Finite-State Transducers
# Figure 3.6: Examples of original and resulting WFSTs for rational operations
# but we use prob Semiring here
let 
    T = WFST();
    addarc(T, "0 1 a:A/1")
    addarc(T, "0 2 b:C/1.5")
    addarc(T, "1 3 c:B/0.3")
    addarc(T, "2 3 d:D/2")
    addstart(T, 0, 0.2)
    addfinal(T, 3, 0.5)

    G = WFST();
    addarc(G, "0 1 d:F/0.2")
    addarc(G, "1 1 e:E/1.2")
    addarc(G, "1 2 f:D/0.6")
    addstart(G, 0, 1.3)
    addfinal(G, 2, 0.8)

    setdpi(50)
    showfsts([ T,  G,  T+G,  T*G ],
             labels=["T","G","T+G","T*G"], boxcolor="lightgray")
end
```

```julia
# Mehryar Mohri, Weighted Automata Algorithms : Fig. 4
let
    T = WFST();
    addarc(T, "0 0 b:a/.1")
    addarc(T, "0 1 b:a/.2")
    addarc(T, "0 1 a:a/.3")
    addarc(T, "1 1 b:a/.4")
    addstart(T, 0)
    addfinal(T, 1, 0.5)

    G = WFST();
    addarc(G, "0 1 a:b/.1")
    addarc(G, "1 0 b:b/.3")
    addarc(G, "1 1 b:a/.2")
    addarc(G, "1 2 a:a/.4")
    addarc(G, "2 1 b:a/.5")
    addarc(G, "2 2 a:b/.6")
    addstart(G, 0)
    addfinal(G, 2, 0.7)

    showfsts([T,G,T+G,T*G, star(T)],
             labels=["T","G","T+G","T*G", "star(T)"],
             boxcolor="lightgray")
end
```
