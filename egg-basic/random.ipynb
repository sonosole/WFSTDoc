# Random Generation

````julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
setdpi(50)
````

## Tree Like WFST
### API
```julia
    treewfst(Σ::Set{I},
             Δ::Set{O},
              ::Type{W};
        wrange::StepRangeLen=0.0 : 1e-2 : 1.0,
            NS::Int=20) where{I,O,W} -> T::WFST{Int, I, O, W}
```
Generate tree like WFST randomly.
+ `Σ` is input symbol set
+ `Δ` is output symbol set
+ `W` is the semiring type
+ `wrange` is the sampling range of weights
+ `NS` is the number of states

````julia
let
    Isyms = Set('a':'z')
    Osyms = Set('A':'Z')
    Wtype = Float32
    M = onefinal(treewfst(Isyms, Osyms, Wtype, NS=17))
    pshow(leveldot(M))
end
````

## Totally Random WFST
### API
```JULIA
    randwfst(Σ::Set{I},
             Δ::Set{O},
              ::Type{W};
              wrange=0.0 : 1e-2 : 1.0,
              outdegrees::Int=3,
              NI::Int=1,
              NF::Int=1,
              NS::Int=20) where{I,O,W} -> T::WFST{Int, I, O, W}
```
Generate a rand WFST object. `T` probably has many unsuccessful paths 😂
+ `Σ` is input symbol set
+ `Δ` is output symbol set
+ `W` is the semiring type
+ `wrange` is the sampling range of weights
+ `outdegrees` is the average out degrees of each node
+ `NI` is the number of inital states
+ `NF` is the number of final states
+ `NS` is the number of states

````julia
let
    Isyms = Set('a':'z')
    Osyms = Set('A':'Z')
    Wtype = Float32
    M = randwfst(Isyms, Osyms, Wtype, NS=10)
    pshow(dot(M))
end
````
