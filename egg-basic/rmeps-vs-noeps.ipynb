# rmeps vs noeps

```julia
using Dioids
using ShowDOT
using WeightedFST
showfsts = pshow ∘ dots;
```

```julia
let W = Dioid{min,+,Float64}

    T = WFST{Int,String,String,W}()
    addarc(T, "0 3 A:a/0.7")
    addarc(T, "3 7 :/0.0")   # ε:ε
    addarc(T, "7 4 B:b/0.1")
    addarc(T, "7 5 C:c/0.2")
    addarc(T, "7 6 F:f/0.6")
    addarc(T, "4 8 :/0.0")   # ε:ε
    addarc(T, "5 9 :/0.2")   # ε:ε
    addstart(T, 0)
    addfinal(T, 6, 0.7)
    addfinal(T, 8, 0.2)
    addfinal(T, 9, 0.1)

    #= rmeps(T) has some redundant states or arcs
    while noeps(T) has no useless states or arcs =#
    setdpi(50)
    showfsts([T,   rmeps(T),   connect(rmeps(T)),  noeps(T)],
             labels=["T", "rmeps(T)", "connect(rmeps(T))", "noeps(T)"],
             boxcolor="lightgray")
end
```
