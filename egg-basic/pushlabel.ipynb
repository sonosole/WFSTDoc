# Push Label to Start or Final

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

```julia
T = WFST()
addpath(T, "0 a 0 c: 1 b:d 0")
addpath(T, "1 a: 2 b: 3 c:d 4 d: 2")
addpath(T, "4 a:d 4 a: 5 f:d 6")
addstart(T, 0)
addfinal(T, 6)
setdpi(60)
showfst(T, label="Origin")
```

push output label to start

```julia
G = pusho(deepcopy(T), "tostart")
showfst(G, label="push olabel to statr")
```

push output label to final

```julia
R = reverse(T)
H = pusho(deepcopy(R), "tofinal")
pshow(depthdot(H, label="push olabel to final"))
```

```julia
let T = WFST{Int,String,String,Int}()
    addarc(T, "0 1 :d")
    addarc(T, "1 2 :abc")
    addarc(T, "1 2 :ab")
    addstart(T, 0)
    addfinal(T, 2)
    S = swapio(T)

    # check pusho and pushi
    G = pusho(deepcopy(T), "tostart")
    H = pushi(deepcopy(S), "tostart")
    showfsts([T, G, S, H],
             labels = ["T", "push olabel to start", "invert(T)", "push ilabel to start"],
             keepone = false,
             boxcolor = "lightgray",
             edgefont = "Georgia")    
end
```

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int, String, String, W}()
    addpath(T, "0 a:b/0.1 1 :b/0.2 2 :c/0.3 3")
    addpath(T, "1 :f/0.3 3")
    addstart(T, 0)
    addfinal(T, 3, 0.5)
    I = deepcopy(T)
    F = deepcopy(T)
    showfsts([pusho(F,"tofinal"), T, pusho(I,"tostart")],
             labels = ["push olabel to final", "T", "push olabel to start"],
             boxcolor="grey",
             bgcolor="lightgrey")
end
```

```julia
let W = Dioid{min,+,Float64}
    T = WFST{Int, String, String, W}()
    addpath(T, "0 a:b/0.1 1 b:/0.2 2 c:/0.3 3")
    addpath(T, "1 f:/0.3 3")
    addstart(T, 0)
    addfinal(T, 3, 0.5)
    I = deepcopy(T)
    F = deepcopy(T)
    showfsts([pushi(F,"tofinal"), T, pushi(I,"tostart")],
             labels = ["push ilabel to final", "T", "push ilabel to start"],
             boxcolor="grey",
             bgcolor="lightblue")
end
```
