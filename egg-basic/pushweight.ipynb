# Push Weight
**Principle**: Redistributes weights along paths based on semiring arithmetic and a potential function; it is an equivalent transformation that preserves the total weight of all paths.

**Formula**: Given a potential function $ϕ: Q \to K$, the transformed arc weights, initial weights and final weights satisfy:
$$
\begin{align*}
w'(e) &= ϕ(p)^{-1} ⊙ w(e) ⊙ ϕ(q) \ \ \ ∀ e = (p,a,b,w,q) \\
λ'(q) &= λ(q) ⊙ ϕ(q) \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \  ∀ q ∈ I \\
ρ'(q) &= ϕ(q)^{-1} ⊙ ρ(q)  \ \ \  \ \ \ \ \ \ \ \ \ \ \ \ \ \ ∀ q ∈ F
\end{align*}
$$

**Purpose**: Unifies weight distribution, avoids numerical drift, and improves stability of subsequent determinization and minimization.

```julia
# Use semiring pkg and showing pkg
using Dioids
using ShowDOT
using WeightedFST
showfst  = pshow ∘ dot;
showfsts = pshow ∘ dots;
```

example from paper *A Weight Pushing Algorithm for Large Vocabulary Speech Recognition, Mehryar Mohri, Michael Riley*

```julia
HardMin = Dioid{min,+,Int}          # Tropical Semiring
SoftMin = Dioid{nlogadd,+,Float64}  # Log Semiring

T = WFST{Int,String,String,SoftMin}
A = WFST{Int,String,String,HardMin}()

addpath(A, "0 a/0 1 e/0 3")
addpath(A, "0 b/1 1 f/1 3")
addarc(A, "0 1 c/4")
addpath(A, "0 d/0 2 e/10 3")
addpath(A, "0 e/1 2 f/11 3")
addstart(A, 0)
addfinal(A, 3)
showfst(A)
```

push weight to start state

```julia
a = deepcopy(A)
ATrop = pushw(a, "tostart")             # push A over Tropical Semiring
ALog = pushw(convert(T, a), "tostart")  # push A over Log Semiring
setdpi(50)
showfsts([A, ALog, ATrop], keepsame=false, digits=3,
         labels=["origin", "push-over-log-semiring", "push-over-tropical-semiring"])
```

push weight to final state

```julia
a = deepcopy(A)
ATrop = pushw(a, "tofinal")             # push A over Tropical Semiring
ALog = pushw(convert(T, a), "tofinal")  # push A over Log Semiring
showfsts([A, ALog, ATrop],keepsame=false, digits=3,
    labels=["origin", "push-over-log-semiring", "push-over-tropical-semiring"])
```
